# Audience exchange voting studio

For audience stewards and moderators: exercise the local SQLite exchange with clearly fictional labels and two illustrative candidates. The exchange stores one vote per label and candidate and applies a declared quorum/tie-break policy; it cannot establish unique people, eligibility, or resistance to coordinated voting.

## Cast illustrative local votes

This in-memory example leaves no database file. Voter labels are not identities or scholarly signatures. A quorum result is a count under the local policy, not an endorsement of factual claims.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / 'src').is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))

from crowd_cinema.exchange import LocalExchange

with LocalExchange() as exchange:
    exchange.create_slot('audience-demo', quorum=2)
    exchange.submit_take('audience-demo', 'candidate-a', 'demo-maker-a', 'Illustrative cut A')
    exchange.submit_take('audience-demo', 'candidate-b', 'demo-maker-b', 'Illustrative cut B')
    exchange.cast_vote('audience-demo', 'candidate-a', 'demo-voter-1')
    exchange.cast_vote('audience-demo', 'candidate-b', 'demo-voter-1')
    exchange.cast_vote('audience-demo', 'candidate-a', 'demo-voter-2')
    result = exchange.tally_slot('audience-demo')
result
```

## Read the result carefully

The tally's winner is deterministic for this small local policy and data set. Quorum counts distinct labels, not verified people; an appeal route and human moderation remain important. Never publish this demonstration as a real poll.
